## A. Python 基础

### 1. 变量：给一个值起名字

变量可以保存字符串、整数、小数或布尔值。`=` 是赋值；`print` 用来观察结果。先区分数值 `30` 和字符串 `"30"`。

````python
learner = "小林"
study_minutes = 30
learning_rate = 0.01
is_beginner = True

print("学习者：", learner)
print("学习分钟数：", study_minutes, "类型：", type(study_minutes).__name__)
print("学习率：", learning_rate, "是否初学：", is_beginner)
print(f"你好，{learner}，今天计划学习 {study_minutes} 分钟。")

# 思考：30 与 "30" 的类型相同吗？把 study_minutes 改为 45 后再运行。
````

### 2. 条件判断：根据条件执行不同分支，缩进表示代码归属

`if / elif / else` 根据条件选择一个分支。比较运算符 `>=` 表示大于等于。Python 用缩进组织代码块；通常每层缩进四个空格。

````python
score = 82
if score >= 90:
    level = "非常熟练"
elif score >= 60:
    level = "继续练习"
else:
    level = "先回顾示例"
print("当前建议：", level)
````

### 3. 列表与循环：保存多个值，并逐个处理

列表按顺序保存多个元素，下标从 0 开始。`for` 遍历元素，`while` 在条件成立时重复执行。`enumerate` 同时给出下标和值；列表推导式是创建列表的一种紧凑写法。

````python
scores = [70, 85, 92]
scores.append(88)
print("第一个分数：", scores[0])  # 下标从 0 开始
print("最后一个分数：", scores[-1])
print("前两个分数：", scores[:2])  # 切片右端不包含在内

total = 0
for index, value in enumerate(scores):
    total += value
    print(f"第 {index + 1} 次练习：{value}")
print("平均分：", total / len(scores))

countdown = 3
while countdown > 0:
    print("剩余练习轮数：", countdown)
    countdown -= 1  # 如果遗漏更新，循环就可能无法结束

passed_scores = [value for value in scores if value >= 80]
print("至少 80 分的成绩：", passed_scores)
print("range(3) 生成：", list(range(3)))
````

### 4. 字典：用键查找对应的值

字典用键找到值。列表适合按位置访问，字典适合按名称访问；两者可以互相嵌套。

````python
record = {"name": learner, "stage": 1, "scores": scores}
record["finished"] = False
print("学习阶段：", record["stage"])
print("缺少的键可以使用默认值：", record.get("note", "暂无备注"))
for key, value in record.items():
    print(key, "->", value)
````

### 5. 函数：输入参数，返回结果

`def` 定义函数，参数接收输入，`return` 返回结果。打印与返回是不同的：打印用于观察，返回的值可以交给后续代码继续运算。异常用来报告无法正常完成的操作。

````python
def average(values):
    """计算一个非空数值列表的平均值。"""
    if len(values) == 0:
        raise ValueError("不能计算空列表的平均值")
    return sum(values) / len(values)


print("用函数算平均分：", average(scores))

# 输入不符合约定时，可以抛出异常；调用方可以捕获并处理。
try:
    average([])
except ValueError as error:
    print("捕获到预期的输入错误：", error)
````

### 6. 类：把数据和操作数据的方法放在一起

类描述一类对象的数据和行为。`__init__` 初始化新对象，`self` 表示当前对象。创建两个对象，观察它们各自的主题列表是否独立。以后搭建 PyTorch 模型时，也会用到类。

````python
class StudyProgress:
    """学习进度；self 指向当前创建的对象。"""

    def __init__(self, name):
        self.name = name
        self.completed_topics = []  # 每个对象拥有自己的列表

    def complete(self, topic):
        self.completed_topics.append(topic)

    def summary(self):
        return f"{self.name} 已完成 {len(self.completed_topics)} 个主题"


progress = StudyProgress(learner)
progress.complete("变量与循环")
progress.complete("函数")
print(progress.summary())
print("已完成的主题：", progress.completed_topics)
````

### 7. 文件读写：把内存中的数据保存到磁盘，再读回来

内存变量在程序结束后不会自动保存。这里用 `pathlib` 管理路径，用 `with` 自动关闭文件，用 JSON 保存简单数据。运行会生成演示文件 `data/learning_record.json`。

````python
import json
from pathlib import Path

# 脚本使用自身所在目录；Notebook 放在同一目录，通常以该目录作为工作目录。
lesson_dir = Path(__file__).resolve().parent if "__file__" in globals() else Path.cwd()
# 也支持从项目根目录启动 Notebook 内核。
if (lesson_dir / "stage_01_python").is_dir():
    lesson_dir = lesson_dir / "stage_01_python"
data_dir = lesson_dir / "data"
data_dir.mkdir(exist_ok=True)
record_path = data_dir / "learning_record.json"

saved_record = {"name": progress.name, "completed_topics": progress.completed_topics}
with record_path.open("w", encoding="utf-8") as file:
    json.dump(saved_record, file, ensure_ascii=False, indent=2)

with record_path.open("r", encoding="utf-8") as file:
    loaded_record = json.load(file)
print("从文件读回的数据：", loaded_record)
print("文件位置：", record_path.resolve())

# 试一试：增加一个 completed_topics 项，再写入并读取，观察 JSON 文件变化。
````

## B. NumPy 数组

### 1. 数组和列表的区别

NumPy 数组适合批量数值运算。`shape` 描述每个轴的长度；一维数组形状 `(3,)` 中的逗号表示这是只有一个元素的元组。先预测下面两个乘法的结果。

````python
import numpy as np

python_list = [1, 2, 3]
vector = np.array([1, 2, 3], dtype=np.float32)
print("列表 * 2：", python_list * 2)  # 重复列表
print("数组 * 2：", vector * 2)  # 每个元素乘 2
print("向量形状：", vector.shape, "维数：", vector.ndim, "数据类型：", vector.dtype)
````

### 2. 二维数组：轴 0 是行，轴 1 是列

二维数组可以看作一张表。形状 `(2, 3)` 表示 2 行 3 列。整数索引通常会去掉一个轴，切片可以保留这个轴；这会影响后续运算是否匹配。

````python
x = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.float32)
print("二维数组：\n", x)
print("shape：", x.shape, "ndim：", x.ndim, "size：", x.size)
print("第 0 行：", x[0], "shape：", x[0].shape)
print("保留二维结构的第 0 行：", x[0:1], "shape：", x[0:1].shape)
print("第 1 列：", x[:, 1])
print("第 1 行第 2 列：", x[1, 2])
````

### 3. 创建数组、改变形状、转置

`reshape` 按新的形状组织元素，元素总数保持不变。二维数组的 `.T` 交换行与列。请先预测 `(2, 3)` 转置后的形状。

````python
print("全零数组：\n", np.zeros((2, 3), dtype=np.float32))
print("全一数组：\n", np.ones((2, 3), dtype=np.float32))
flat = np.arange(6)
reshaped = flat.reshape(2, 3)  # 元素总数必须相同：6 = 2 * 3
print("原数组：", flat, "reshape 后：\n", reshaped)
print("x 转置后：\n", x.T, "shape：", x.T.shape)
````

### 4. 基本运算、按轴求平均、广播

`axis` 指明沿哪个轴汇总。`axis=0` 把各行汇总，留下每列的统计值；`axis=1` 把各列汇总，留下每行的统计值。广播让兼容形状的数组参与运算，下面将同一个偏置加到每一行。

````python
print("每个元素加 10：\n", x + 10)
print("每个元素平方：\n", x ** 2)
print("所有元素的平均值：", x.mean())
print("每列平均值（沿轴 0 汇总）：", x.mean(axis=0))
print("每行平均值（沿轴 1 汇总）：", x.mean(axis=1))

bias = np.array([10, 20, 30], dtype=np.float32)
print("x + bias：\n", x + bias)
# (2, 3) + (3,)：从最后一个维度对齐，bias 自动应用于每一行。
# 广播要求对齐的每个维度大小相等，或者其中一个为 1。
````

### 5. 区分逐元素乘法 * 和矩阵乘法 @

`*` 对应位置相乘；`@` 做矩阵乘法。对于二维数组，`(m, k) @ (k, n)` 的结果是 `(m, n)`。请手算输出的四个数，再运行验证。

````python
print("逐元素乘法 x * x：\n", x * x, "shape：", (x * x).shape)
weights = np.array([[1, 0], [0, 1], [1, 1]], dtype=np.float32)
output = x @ weights
print("矩阵乘法 x @ weights：\n", output)
print("形状变化：", x.shape, "@", weights.shape, "->", output.shape)
# (2, 3) @ (3, 2) -> (2, 2)，中间的 3 必须相同。
# 左上角 = 1*1 + 2*0 + 3*1 = 4；请自己手算其余三个位置。
````

### 6. 批处理：把多个样本放到一个数组中

这里把每行当作一个样本，一次计算多行就是批处理。`batch @ weights + bias` 会在后面的神经网络中反复出现；现在先检查输入和输出的形状。

````python
# 假设每行是一个样本的 3 个特征，一次处理两个样本。
batch = x
output_bias = np.array([0.5, -0.5], dtype=np.float32)
batch_output = batch @ weights + output_bias
print("batch shape：", batch.shape)
print("批处理结果：\n", batch_output, "shape：", batch_output.shape)

# 显式固定随机种子，让本示例的随机数据可重复。
rng = np.random.default_rng(42)
random_features = rng.normal(size=(2, 3)).astype(np.float32)
print("随机特征：\n", random_features)

# 下一阶段会解释矩阵乘法的数学意义；这里先掌握运算符与形状。
````
